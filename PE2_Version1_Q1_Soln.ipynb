**Q1a   5 marks**

- Load the CSV.
- Retain only records where Units > 25.
- Display the retained (**filtered**) records and their row count. Retain all original columns at this stage.

Submit: Python code and printed filtered records and row count in the Q1 box. No file or screenshot.


```python
import pandas as pd
df = pd.read_csv("retail_sales.csv")                # 1 mark
```

```python
filtered = df[df["Units"] > 25].copy()              # 2 marks
print(filtered)                                     # 1 mark
print("Retained rows:", len(filtered))              # 1 mark
```

**Q1b   5 marks**

- Create **Sales = Units × Unit_Price** for the filtered records (from [a]).
- **Select** only Month, Region and Sales into a **separate DataFrame** and display it.

Submit: Python code and the printed three-column DataFrame in the Q1 box. No file or screenshot.


```python
filtered["Sales"] = filtered["Units"] * filtered["Unit_Price"]           # 2 marks
selected = filtered[["Month", "Region", "Sales"]]                        # 2 marks
print(selected)                                                          # 1 mark
```

**Q1c   5 marks**

- Group the selected data (from [b]) by Region and calculate **total Sales for each region**.
- Check that the sum of regional totals equals the sum of Sales in the filtered records;
- Display both totals and state whether they match.

Submit: Python code, regional totals, the two check totals and a match statement in the Q1 box. No output file.


```python
regional = selected.groupby("Region")["Sales"].sum()        # 2 marks
regional_total = regional.sum()                             # 1 mark
filtered_total = filtered["Sales"].sum()                    # 1 mark
print(regional_total, filtered_total)                       # 1 mark
print("Match:", regional_total == filtered_total)
```
